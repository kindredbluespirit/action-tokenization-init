### 3. The Evolution of Action Representations in VLAs


```python
# Timeline (conceptual)
print("2022-2023: Early Discretization")
print("  RT-1, RT-2, OpenVLA: per-dimension binning")
print("  → Simple but lossy, poor for dexterous tasks")

print("\n2023: Continuous Representations")
print("  ACT: action chunks via CVAE")
print("  Diffusion Policy: iterative denoising")
print("  → Better dexterity, but no language conditioning")

print("\n2024-2025: VLAs with Continuous Actions")
print("  pi0: flow matching + action expert")
print("  SmolVLA: cross-attention action expert")
print("  → Language + vision + continuous actions")

print("\n2025: Tokenization Experiment")
print("  pi0-FAST: DCT+BPE tokens, autoregressive")
print("  → 5× faster training, but slow inference")

print("\n2025-2026: Back to Continuous")
print("  pi0.5: flow matching + action expert (again)")
print("  → Tokenization promising but not ready for production")
```

### 4. Why Physical Intelligence Went Back to Action Expert


```python
print("Possible reasons pi0.5 dropped pure tokenization:")
print("  1. Autoregressive inference too slow for real-time control")
print("  2. Quantization error from DCT+BPE loses fine dexterity")
print("  3. Flow matching produces smoother action trajectories")
print("  4. Action expert specializes for control; shared backbone doesn't")
print("  5. KV-caching helps but doesn't solve sequential bottleneck")
```

### 5. Open Questions

The field is still evolving. Recent work (2025-2026) explores:

- **OAT** (Ordered Action Tokenization): learned tokenizers with causal ordering
- **QueST**: VQ-VAE style action compression
- **BEAST**: B-spline encoded action sequences

The debate — continuous vs discrete actions — is not settled. But the trend in production VLAs is clear: continuous with action experts.


### 6. Final Summary

| Approach | Models | Pros | Cons |
|----------|--------|------|------|
| Continuous + CVAE | ACT | Fast, simple | No language conditioning |
| Continuous + diffusion | Diffusion Policy | Smooth, multimodal | Slow inference |
| Continuous + flow matching | pi0, pi0.5 | Dexterous, fast inference | Slow training |
| Continuous + cross-attn | SmolVLA | Efficient, modular | Less dexterity |
| Discrete tokens (FAST) | pi0-FAST | Fast training | Slow inference, quantization |

**Bottom line**: 5 out of 6 models in leRobot use continuous actions with dedicated action experts. Tokenization remains an active research direction but hasn't replaced continuous methods in production VLAs.
